Import the libraries and create the dataset

```python
# Import pandas for working with tables
import pandas as pd
```

```python

# Import the function used to split the data
from sklearn.model_selection import train_test_split

# Import tools for filling missing values, encoding and scaling
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Import tools that allow us to combine preprocessing steps
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Create a small student performance dataset
data = {
    "Study_hours": [2, 3, 5, 6, 4, 8, 7, 1, 5, 9, 3, 6],
    "Attendance": [55, 60, 75, 80, None, 95, 88, 50, 78, 98, 65, 85],
    "Age": [18, 19, 20, 21, 19, 22, 20, 18, 21, 23, 19, 20],
    "Study_method": ["Group", "Alone", "Group", "Alone", "Group", "Alone", "Group", "Alone", None, "Group", "Alone", "Group"],
    "Passed": ["No", "No", "Yes", "Yes", "No", "Yes", "Yes", "No", "Yes", "Yes", "No", "Yes"]
}

# Turn the dictionary into a pandas DataFrame
df = pd.DataFrame(data)

# Display the dataset
display(df)
```

Step 2; Explore the dataset

```python
# Show the number of rows and columns
print("Dataset shape:", df.shape)
```

```python
# Show column names, data types and non-missing counts
df.info()
```

```python
# Count missing values in each column
print("\nMissing values:")
print(df.isnull().sum())
```

```python

# Show a statistical summary of numerical columns
display(df.describe())
```

Step 3: Separate the features and target
- The features (X) are the columns the model can use to make a prediction. The target (y) is what we want the model to predict. Here, the target is Passed. I remove the target from the feature columns so it is not accidentally used as an input.

```python
# Select the input columns
X = df[["Study_hours", "Attendance", "Age", "Study_method"]]

# Select the answer the model should predict
y = df["Passed"]

# Display the features and target
print("Features:")
display(X.head())

print("Target:")
display(y.head())
```

Step 4: Split the data into training and test sets

```python
 # Split the features and target into training and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X,                 # Input features
    y,                 # Target to predict
    test_size=0.2,     # Keep 20% for testing
    random_state=42,   # Make the split repeatable
    stratify=y         # Keep target proportions similar
)

# Display the size of each set
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```
